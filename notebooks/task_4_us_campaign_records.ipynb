```python
import pandas as pd

df = pd.read_csv(
    "../data/DSI_kickstarterscrape_dataset.csv",
    encoding="latin1"
)

print("Dataset shape:", df.shape)
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nFirst 20 location values:")
print(df["location"].head(20))

print("\nMissing location values:")
print(df["location"].isna().sum())
```

```python
print(df["location"].dropna().sample(20, random_state=42).to_string(index=False))
```

```python
us_states = {
    "AL", "AK", "AZ", "AR", "CA", "CO", "CT", "DE",
    "FL", "GA", "HI", "ID", "IL", "IN", "IA", "KS",
    "KY", "LA", "ME", "MD", "MA", "MI", "MN", "MS",
    "MO", "MT", "NE", "NV", "NH", "NJ", "NM", "NY",
    "NC", "ND", "OH", "OK", "OR", "PA", "RI", "SC",
    "SD", "TN", "TX", "UT", "VT", "VA", "WA", "WV",
    "WI", "WY", "DC"
}

location_state = (
    df["location"]
    .fillna("")
    .str.rsplit(",", n=1)
    .str[-1]
    .str.strip()
)

us_df = df[location_state.isin(us_states)].copy()

print("Original records:", len(df))
print("U.S. records:", len(us_df))
print("Records not identified as U.S.:", len(df) - len(us_df))
```

```python
print("U.S. dataset shape:", us_df.shape)

print("\nUnique U.S. location values:")
print(us_df["location"].nunique())

print("\nSample of U.S. locations:")
print(us_df["location"].sample(20, random_state=42).to_string(index=False))
```

```python
# Verify that all filtered locations end with a valid U.S. state/DC abbreviation

us_location_states = (
    us_df["location"]
    .str.rsplit(",", n=1)
    .str[-1]
    .str.strip()
)

invalid_states = sorted(set(us_location_states) - us_states)

print("Invalid state values:", invalid_states)
print("Number of invalid state values:", len(invalid_states))
```

```python
print("Sample of U.S.-based campaigns:")
print(us_df[["name", "location", "status"]].head(20).to_string(index=False))
```

## Task 4: Prepare U.S.-Based Campaign Records

### Location Review

The dataset contains 45,957 campaign records and 17 columns. The `location` column was reviewed to understand how campaign locations are represented. Location values generally follow a `City, State/Country` format, such as `Los Angeles, CA` and `Melbourne, Australia`.

There are 1,322 records with missing location values. These records were not included in the U.S.-based subset because a U.S. location could not be determined from the available location information.

### U.S.-Based Campaign Filtering

To identify U.S.-based campaigns, the final component of each location value was extracted and compared against a list containing the 50 U.S. state abbreviations and Washington, D.C.

The filtered dataset contains:

- **Original dataset:** 45,957 records
- **U.S.-based campaigns:** 42,268 records
- **Records not identified as U.S.-based:** 3,689 records
- **Unique U.S. location values:** 4,035

The resulting `us_df` DataFrame contains only campaign records whose location ends with a recognized U.S. state or Washington, D.C. abbreviation.

### Location Information Review

A sample of the filtered records was reviewed using the campaign name, location, and status fields. The sample showed U.S.-based locations including Columbia, MO; Los Angeles, CA; Chicago, IL; New York, NY; and Austin, TX.

The original `location` field was retained rather than creating duplicate city and state columns. This preserves the original location information for potential use in future exploratory analysis.

### Verification

The U.S.-based dataset was verified by checking the state/DC portion of each retained location against the list of recognized U.S. state and D.C. abbreviations.

**Verification result:**

- **Invalid state values:** 0
- **U.S. dataset shape:** 42,268 rows × 17 columns

The resulting U.S.-based campaign dataset is prepared for future exploratory analysis and machine learning.
